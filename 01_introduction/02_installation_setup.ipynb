# 1.2 Installation & Environment Setup


## 📦 Installing spaCy

spaCy is compatible with 64-bit CPython 3.8+ and runs on Unix/Linux, macOS/OS X and Windows. The latest spaCy releases are available over pip and conda.

### Standard Installation
If you haven't installed it yet, you can run:
```bash
pip install -U pip setuptools wheel
pip install -U spacy
```

### Virtual Environments
It is strongly recommended to install spaCy in a virtual environment (like `venv` or `conda` environments). This ensures that your project's dependencies don't conflict with other projects.

### Hardware Acceleration (GPU)
If you are processing massive amounts of text or training complex neural network models, you might want to use a GPU. spaCy supports GPU processing via the `thinc` machine learning library.

To install spaCy with GPU support, you need to append the CUDA version you are using. For example, for CUDA 11.x:
```bash
pip install -U spacy[cuda11x]
```
