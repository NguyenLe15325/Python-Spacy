## 🎬 Processing Text with the `nlp` Object

When you call the `nlp` object on a string of text, spaCy returns a processed `Doc` object. 
This single step runs the entire pipeline (tokenization, tagging, parsing, NER, etc.) under the hood!


```python
import spacy

# Load the small English model
nlp = spacy.load("en_core_web_sm")

# Process a sentence
text = "Apple is looking at buying U.K. startup for $1 billion."
doc = nlp(text)

print(f"Successfully processed document of length: {len(doc)} tokens")
```

## 🧩 Exploring Tokens

A `Doc` is a sequence of `Token` objects. We can iterate over the doc to see what spaCy learned about each word.

Here are some useful Token attributes:
- `.text`: The original word text.
- `.lemma_`: The base form of the word.
- `.pos_`: The simple UPOS part-of-speech tag (Noun, Verb, Adjective).
- `.dep_`: Syntactic dependency (how words relate to each other).
- `.is_stop`: Is the word a common stop word?


```python
import pandas as pd

# Create a list of dictionaries to format nicely in a Pandas DataFrame
token_data = []
for token in doc:
    token_data.append({
        "Text": token.text,
        "Lemma": token.lemma_,
        "POS": token.pos_,
        "Dependency": token.dep_,
        "Is Alpha": token.is_alpha,
        "Is Stopword": token.is_stop
    })

# Display as a table
pd.DataFrame(token_data)
```

## 🏷️ Named Entity Recognition (NER)

A named entity is a "real-world object" that's assigned a name – for example, a person, a country, a product, or a book title. spaCy can recognize various types of named entities in a document.

You can access these via the `doc.ents` property. Let's see what spaCy found in our sentence!


```python
for ent in doc.ents:
    # Print the entity text and its label
    print(f"{ent.text:<15} | {ent.label_:<10} | {spacy.explain(ent.label_)}")
```

Notice how spaCy provides the `spacy.explain()` helper function? If you ever see a tag or label you don't understand, just pass it to `spacy.explain()` to get a human-readable description!


## 📝 Sentence Segmentation

spaCy also automatically splits your document into sentences based on linguistic logic (not just splitting on periods).
Let's process a multi-sentence text.


```python
long_text = "This is sentence one. And here is sentence two! What about sentence three? Yes, it works."
long_doc = nlp(long_text)

print("Sentences found:")
for i, sentence in enumerate(long_doc.sents):
    print(f"{i+1}: {sentence.text}")
```
