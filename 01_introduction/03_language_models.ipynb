# 1.3 Language Models (Trained Pipelines)


## 🧠 What is a Trained Pipeline (Model)?

To do anything useful with spaCy, you need a trained pipeline. These pipelines are pre-trained on large datasets (like Wikipedia, news articles, or web text) and contain:
- **Vocabulary**: Words the model knows.
- **Weights**: Statistical information to make predictions (e.g., predicting that "Apple" is a company, not a fruit, based on context).
- **Rules**: Language-specific rules for tokenization, lemmatization, etc.

spaCy provides models for many different languages (English, Spanish, German, Chinese, etc.).


## 📏 Model Sizes and Variations

For most languages, spaCy offers three main sizes of statistical models, plus a state-of-the-art transformer model:

1. **`sm` (Small)**: ~12 MB. Contains no word vectors. Very fast loading and processing. Ideal for basic prototyping or when memory is strictly limited.
2. **`md` (Medium)**: ~40 MB. Includes a reduced word vector table (e.g., 20k unique vectors). Good balance between size, speed, and accuracy.
3. **`lg` (Large)**: ~560 MB. Includes a large word vector table (e.g., 500k unique vectors). Best accuracy for standard pipelines. Excellent for tasks requiring deep semantic similarity.
4. **`trf` (Transformer)**: ~400+ MB. Uses a transformer architecture (like BERT or RoBERTa). State-of-the-art accuracy, but significantly slower and requires more memory (GPU recommended).

### Naming Convention
Models follow a strict naming convention: `[lang]_[type]_[genre]_[size]`
- Example: `en_core_web_sm`
  - `en`: English
  - `core`: Core NLP tasks (tagging, parsing, NER)
  - `web`: Trained on web text (blogs, news, comments)
  - `sm`: Small size


## ⬇️ Downloading a Model

Usually, you download models via the command line:
```bash
python -m spacy download en_core_web_sm
```

Let's download the small English model directly from this notebook using spaCy's download module.


````python
import spacy
from spacy.cli import download

print("Downloading en_core_web_sm...")
# This is equivalent to running `python -m spacy download en_core_web_sm`
download("en_core_web_sm")
print("Download complete!")
````

## 🏗️ Loading the Model

Once a model is downloaded, you load it into an `nlp` object. This object contains the processing pipeline and language-specific rules.


````python
import spacy

# Load the downloaded model
nlp = spacy.load("en_core_web_sm")

print("Pipeline loaded successfully!")
print(f"Pipeline name: {nlp.meta['name']}")
print(f"Pipeline language: {nlp.meta['lang']}")

# What components are inside this pipeline?
print("\nPipeline Components:")
for name, component in nlp.pipeline:
    print(f"- {name}")
````
