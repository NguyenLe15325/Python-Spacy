# 1.5 The spaCy Architecture


## 🏛️ High-Level Architecture

Understanding spaCy's central data structures will save you a lot of time and help you write much faster, more efficient code.

### The Core Objects

1. **`Language` / `nlp`**: The central coordinator. It holds the vocabulary, pipeline components, and language-specific rules. It is responsible for turning raw text into a `Doc`.
2. **`Doc`**: A container for accessing linguistic annotations. It is created by the `nlp` object and holds a sequence of `Token` objects.
3. **`Token`**: An individual word, punctuation mark, or symbol within a `Doc`.
4. **`Span`**: A continuous slice of tokens from a `Doc`. (e.g., an entity like "U.K." or a phrase like "is looking at").


```python
import spacy
nlp = spacy.load("en_core_web_sm")

doc = nlp("I love natural language processing.")

# Exploring types
print(f"type(nlp)   = {type(nlp)}")
print(f"type(doc)   = {type(doc)}")
print(f"type(token) = {type(doc[0])}")
print(f"type(span)  = {type(doc[2:5])}")
```

## 💾 Memory Management: Vocab, StringStore, and Lexemes

spaCy is famous for its memory efficiency. If you have a document with 100,000 words, and the word "coffee" appears 5,000 times, spaCy does *not* save the string "coffee" 5,000 times in memory.

Instead, spaCy hashes all strings to **64-bit integers**.

### The `StringStore`
The `StringStore` acts as a lookup table that works in both directions:
- `String` -> `Hash`
- `Hash` -> `String`

The `StringStore` is shared across the entire `nlp` object, meaning all documents processed by that object share the same vocabulary.


```python
# Let's look at the hashing in action
coffee_hash = nlp.vocab.strings["coffee"]
coffee_string = nlp.vocab.strings[coffee_hash]

print(f"String: '{coffee_string}'")
print(f"Hash:   {coffee_hash}")

# It works for ANY string, even ones spaCy hasn't seen before
custom_hash = nlp.vocab.strings["Supercalifragilisticexpialidocious"]
print(f"\nCustom String Hash: {custom_hash}")
```

### The `Lexeme` Object

While a `Token` represents a word *in context* (e.g., the word "booking" could be a verb or a noun depending on the sentence), a **`Lexeme`** represents a word *out of context*.

A `Lexeme` contains information that is true for the word no matter where it appears (e.g., is it alphabetical? is it a number? what is its shape?). 

You get Lexemes directly from the `nlp.vocab`.


```python
# Let's get a Lexeme from the vocabulary
lexeme = nlp.vocab["coffee"]

print(f"Lexeme text: {lexeme.text}")
print(f"Lexeme hash: {lexeme.orth}")
print(f"Is alpha?  : {lexeme.is_alpha}")
print(f"Is title?  : {lexeme.is_title}")
print(f"Word Shape : {lexeme.shape_}")  # 'xxxx' means all lowercase letters
```

## 🔄 The Pipeline Data Flow

When you call `nlp(text)`, here is what happens sequentially:

1. **Tokenizer**: The raw text is split into a `Doc` of `Token` objects. (This is the only component that modifies the document structure).
2. **Tagger**: Assigns Part-of-Speech tags.
3. **Parser**: Assigns dependency labels and parses the syntactic structure. Also detects sentence boundaries.
4. **NER**: Detects and labels named entities.
5. **Lemmatizer**: Assigns base forms based on the POS tags.
6. **Attribute Ruler / Morphologizer**: Refines tags and morphological features.

*Note: The exact pipeline components depend on the model you load.*
