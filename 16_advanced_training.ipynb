## 🏗️ Multi-Component Pipelines

So far, we have generated configs for a single component: `--pipeline ner` or `--pipeline textcat`.

But a real-world model often does multiple things. You can train a model that does POS tagging, NER, AND Text Categorization all at once!

```bash
# Generate a config with multiple components!
python -m spacy init config config.cfg --lang en --pipeline "tagger,parser,ner,textcat" --optimize accuracy
```

When training a multi-component pipeline, the neural network layers can actually *share weights* across the different tasks, making the overall model smarter and faster to execute!

## 🧬 Data Augmentation

What if you only have 50 training sentences, but you know you need 500? You can use **Data Augmentation**.

Data augmentation is the process of slightly modifying your existing training data to create new, valid examples. 
spaCy provides built-in callbacks to automatically augment data during the training loop!

Common techniques:
- **Synonym Replacement**: Replacing words with their synonyms (e.g. "I bought a phone" -> "I purchased a phone").
- **Orthographic Augmentation**: Randomly changing the capitalization of words, or removing punctuation, so the model learns not to rely purely on capital letters to find entities!

## 🌍 Using Pre-Trained Transformers (Domain Adaptation)

In Module 7, we mentioned that static Word Vectors (like Word2Vec) have limitations, and that **Transformers** (like BERT or RoBERTa) are the modern solution.

You can configure spaCy to use a HuggingFace Transformer as the base layer for your model instead of the default CNN!

To do this, you use the `--optimize accuracy` flag and specify the transformer model in your config:

```bash
# Generate a transformer-based config
python -m spacy init config config.cfg --lang en --pipeline ner --optimize accuracy
```

Inside `config.cfg`, you can set `model.name = "roberta-base"`. When spaCy trains, it will automatically download the RoBERTa weights from HuggingFace and **fine-tune** them on your specific NER or TextCat task! 

*Note: Transformer models require a GPU to train and run efficiently!*

## 🧪 MLOps Integrations (Weights & Biases)

When you are training models professionally, you need to track your experiments. Did increasing the batch size improve the F1 score? What was the loss curve?

spaCy integrates natively with MLOps tracking tools like **Weights & Biases (wandb)** and **MLflow**. 

You can simply add a logger block to your `config.cfg`:

```ini
[training.logger]
@loggers = "spacy.WandbLogger.v4"
project_name = "my_ner_project"
```

When you run the `spacy train` command, it will automatically stream real-time charts and metrics to your Weights & Biases dashboard!
