## 🏗️ Multi-Component Training

Most complex NLP systems do not just do one thing. If you are building a legal document parser, you might need to extract `CLAUSE` entities (NER) and classify the document as `NDA` or `CONTRACT` (TextCat).

Instead of training two separate models, you can train a multi-component pipeline!

```bash
# Generate a config with BOTH tok2vec, ner, and textcat
python -m spacy init config config_multi.cfg --lang en --pipeline "tok2vec,ner,textcat" --optimize accuracy
```

**The Benefit of Weight Sharing:**
If you train both components simultaneously, the underlying `tok2vec` (or Transformer) layer receives gradients from *both* the NER head and the TextCat head. This forces the base network to learn richer, more generalized representations of the text!

## 🧬 Data Augmentation

Deep learning models require thousands of examples to prevent overfitting. If you only have 100 sentences, the model will just memorize them.

**Data Augmentation** modifies your existing training data dynamically during training to create new examples. 
spaCy provides a built-in `[corpora.train.augmenter]` block in the `config.cfg`.

**Common Techniques:**
1. **Orthographic Augmentation:** Randomly changes capitalization (e.g., "iPhone" -> "iphone" or "IPHONE"). This forces the model to look at the *context*, not just the capital letters.
2. **Punctuation Stripping:** Randomly removes periods or commas to simulate messy text.
3. **Synonym Replacement (via custom callbacks):** Replacing "purchased" with "bought".

To enable basic orthographic augmentation in `config.cfg`:
```ini
[corpora.train.augmenter]
@augmenters = "spacy.orth_variants.v1"
level = 0.1  # 10% chance to augment
lower = 0.5  # 50% chance to lowercase
```

## 🤖 Active Learning & Prodigy

Data labeling is the most expensive part of training. **Active Learning** is the process of having the model help you label the data.

1. You train a weak model on 50 examples.
2. The model predicts labels on 10,000 unlabeled sentences.
3. The model calculates its **Confidence Score** for each prediction.
4. You sort the sentences and manually review *only the ones where the model was 50% confident* (the hardest ones).
5. You add those corrections to your training set and retrain.

This is the core loop of **Prodigy**, a professional annotation tool built by the creators of spaCy (Explosion AI). Prodigy integrates directly with spaCy models and provides a web interface for exactly this workflow!

## 🧪 MLOps & Experiment Tracking

When tweaking hyperparameters (learning rates, batch sizes, dropout rates), you need to track which config produced the best F1 score.
spaCy integrates natively with MLOps tools like **Weights & Biases (wandb)**.

Instead of reading terminal output, you add a logger to your `config.cfg`:

```ini
[training.logger]
@loggers = "spacy.WandbLogger.v4"
project_name = "support_ticket_router"
run_name = "experiment_01_cnn"
```

When you run `spacy train`, it will automatically stream the Loss curves, precision/recall metrics, and even sample predictions directly to the Weights & Biases web dashboard for visual comparison!
