## 🐛 Debugging spaCy Pipelines

Sometimes things don't work as expected. spaCy has built-in tools to help you figure out what's wrong with your pipeline configuration.

### Inspecting Pipeline Information
You can easily check what a component requires as input, and what it produces as output.


```python
import spacy

nlp = spacy.load("en_core_web_sm")

# Analyze a specific component, like the 'tagger'
print("Tagger Analysis:")
analysis = nlp.analyze_pipes()

# Let's format the output nicely
import pandas as pd
pd.DataFrame(analysis['summary'])
```

In the table above, `assigns` tells you what attributes the component sets (e.g., `token.tag`, `token.pos`). `requires` tells you what must be present on the Doc *before* this component can run.

### Profiling Performance
If your pipeline is running slowly, you can profile it to see which component is taking the most time using `nlp.profile`.


```python
import spacy
nlp = spacy.load("en_core_web_sm")

# We can use the context manager to profile processing time
with nlp.profile() as profile:
    doc = nlp("This is a sentence to profile. " * 100)
    
print("Profiling complete! Time spent in each component:")
for name, data in profile.items():
    print(f"- {name:<10}: {data}")
```
