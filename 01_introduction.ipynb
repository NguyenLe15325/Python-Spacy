# 1.1 What is spaCy?


## 🌟 Introduction to spaCy

spaCy is a free, open-source library for advanced Natural Language Processing (NLP) in Python.

While other NLP libraries were built mainly for academic research and teaching, **spaCy was designed from day one to be used in production.** It helps you build applications that process and "understand" large volumes of text.

### 🚀 Key Philosophies of spaCy:
1. **Opinionated**: spaCy generally provides one highly optimized, state-of-the-art way to do a specific task, saving you from making choices about which algorithm to use.
2. **Fast**: It's written in Cython (C-extension for Python) from the ground up, making its core operations blazingly fast.
3. **Production-Ready**: Models are packaged as easily installable Python modules, making versioning and deployment simple.

---
## ⚖️ spaCy vs NLTK

If you've done NLP in Python before, you've likely used **NLTK** (Natural Language Toolkit).

| Feature | NLTK | spaCy |
|---------|------|-------|
| **Primary Goal** | Teaching and research | Production usage |
| **Algorithms** | Many algorithms for each task | One state-of-the-art algorithm |
| **Performance** | Slower | Highly optimized (Cython) |
| **Deep Learning** | Limited integration | Excellent integration (Transformers, PyTorch, etc.) |

*Think of NLTK as a massive toolbox where you have 10 different types of hammers. spaCy is like a factory assembly line with the perfect automated hammer already set up.*


## 🛠️ What can spaCy do?

Here are some of the core capabilities spaCy offers right out of the box:
- **Tokenization**: Segmenting text into words, punctuations marks, etc.
- **Part-of-speech (POS) Tagging**: Assigning word types to tokens, like verb or noun.
- **Dependency Parsing**: Assigning syntactic dependency labels, describing the relations between individual tokens.
- **Lemmatization**: Assigning the base forms of words (e.g., the lemma of "was" is "be").
- **Named Entity Recognition (NER)**: Labeling named "real-world" objects, like persons, companies or locations.
- **Word Vectors & Similarity**: Comparing words, spans and documents and how similar they are to each other.
- **Text Classification**: Assigning categories or labels to a whole document.

# 1.3 Language Models (Trained Pipelines)


## 🧠 What is a Trained Pipeline (Model)?

To do anything useful with spaCy, you need a trained pipeline. These pipelines are pre-trained on large datasets (like Wikipedia, news articles, or web text) and contain:
- **Vocabulary**: Words the model knows.
- **Weights**: Statistical information to make predictions (e.g., predicting that "Apple" is a company, not a fruit, based on context).
- **Rules**: Language-specific rules for tokenization, lemmatization, etc.

spaCy provides models for many different languages (English, Spanish, German, Chinese, etc.).


## 📏 Model Sizes and Variations

For most languages, spaCy offers three main sizes of statistical models, plus a state-of-the-art transformer model:

1. **`sm` (Small)**: ~12 MB. Contains no word vectors. Very fast loading and processing. Ideal for basic prototyping or when memory is strictly limited.
2. **`md` (Medium)**: ~40 MB. Includes a reduced word vector table (e.g., 20k unique vectors). Good balance between size, speed, and accuracy.
3. **`lg` (Large)**: ~560 MB. Includes a large word vector table (e.g., 500k unique vectors). Best accuracy for standard pipelines. Excellent for tasks requiring deep semantic similarity.
4. **`trf` (Transformer)**: ~400+ MB. Uses a transformer architecture (like BERT or RoBERTa). State-of-the-art accuracy, but significantly slower and requires more memory (GPU recommended).

### Naming Convention
Models follow a strict naming convention: `[lang]_[type]_[genre]_[size]`
- Example: `en_core_web_sm`
  - `en`: English
  - `core`: Core NLP tasks (tagging, parsing, NER)
  - `web`: Trained on web text (blogs, news, comments)
  - `sm`: Small size


## ⬇️ Downloading a Model

Usually, you download models via the command line:
```bash
python -m spacy download en_core_web_sm
```

Let's download the small English model directly from this notebook using spaCy's download module.


````python
import spacy
from spacy.cli import download

print("Downloading en_core_web_sm...")
# This is equivalent to running `python -m spacy download en_core_web_sm`
download("en_core_web_sm")
print("Download complete!")
````

## 🏗️ Loading the Model

Once a model is downloaded, you load it into an `nlp` object. This object contains the processing pipeline and language-specific rules.


````python
import spacy

# Load the downloaded model
nlp = spacy.load("en_core_web_sm")

print("Pipeline loaded successfully!")
print(f"Pipeline name: {nlp.meta['name']}")
print(f"Pipeline language: {nlp.meta['lang']}")

# What components are inside this pipeline?
print("\nPipeline Components:")
for name, component in nlp.pipeline:
    print(f"- {name}")
````

# 1.4 Your First NLP Program


## 🎬 Processing Text with the `nlp` Object

When you call the `nlp` object on a string of text, spaCy returns a processed `Doc` object. 
This single step runs the entire pipeline (tokenization, tagging, parsing, NER, etc.) under the hood!


````python
import spacy

# Load the small English model
nlp = spacy.load("en_core_web_sm")

# Process a sentence
text = "Apple is looking at buying U.K. startup for $1 billion."
doc = nlp(text)

print(f"Successfully processed document of length: {len(doc)} tokens")
````

## 🧩 Exploring Tokens

A `Doc` is a sequence of `Token` objects. We can iterate over the doc to see what spaCy learned about each word.

Here are some useful Token attributes:
- `.text`: The original word text.
- `.lemma_`: The base form of the word.
- `.pos_`: The simple UPOS part-of-speech tag (Noun, Verb, Adjective).
- `.dep_`: Syntactic dependency (how words relate to each other).
- `.is_stop`: Is the word a common stop word?


````python
import pandas as pd

# Create a list of dictionaries to format nicely in a Pandas DataFrame
token_data = []
for token in doc:
    token_data.append({
        "Text": token.text,
        "Lemma": token.lemma_,
        "POS": token.pos_,
        "Dependency": token.dep_,
        "Is Alpha": token.is_alpha,
        "Is Stopword": token.is_stop
    })

# Display as a table
pd.DataFrame(token_data)
````

## 🏷️ Named Entity Recognition (NER)

A named entity is a "real-world object" that's assigned a name – for example, a person, a country, a product, or a book title. spaCy can recognize various types of named entities in a document.

You can access these via the `doc.ents` property. Let's see what spaCy found in our sentence!


````python
for ent in doc.ents:
    # Print the entity text and its label
    print(f"{ent.text:<15} | {ent.label_:<10} | {spacy.explain(ent.label_)}")
````

Notice how spaCy provides the `spacy.explain()` helper function? If you ever see a tag or label you don't understand, just pass it to `spacy.explain()` to get a human-readable description!


## 📝 Sentence Segmentation

spaCy also automatically splits your document into sentences based on linguistic logic (not just splitting on periods).
Let's process a multi-sentence text.


````python
long_text = "This is sentence one. And here is sentence two! What about sentence three? Yes, it works."
long_doc = nlp(long_text)

print("Sentences found:")
for i, sentence in enumerate(long_doc.sents):
    print(f"{i+1}: {sentence.text}")
````

# 1.5 The spaCy Architecture


## 🏛️ High-Level Architecture

Understanding spaCy's central data structures will save you a lot of time and help you write much faster, more efficient code.

### The Core Objects

1. **`Language` / `nlp`**: The central coordinator. It holds the vocabulary, pipeline components, and language-specific rules. It is responsible for turning raw text into a `Doc`.
2. **`Doc`**: A container for accessing linguistic annotations. It is created by the `nlp` object and holds a sequence of `Token` objects.
3. **`Token`**: An individual word, punctuation mark, or symbol within a `Doc`.
4. **`Span`**: A continuous slice of tokens from a `Doc`. (e.g., an entity like "U.K." or a phrase like "is looking at").


````python
import spacy
nlp = spacy.load("en_core_web_sm")

doc = nlp("I love natural language processing.")

# Exploring types
print(f"type(nlp)   = {type(nlp)}")
print(f"type(doc)   = {type(doc)}")
print(f"type(token) = {type(doc[0])}")
print(f"type(span)  = {type(doc[2:5])}")
````

## 💾 Memory Management: Vocab, StringStore, and Lexemes

spaCy is famous for its memory efficiency. If you have a document with 100,000 words, and the word "coffee" appears 5,000 times, spaCy does *not* save the string "coffee" 5,000 times in memory.

Instead, spaCy hashes all strings to **64-bit integers**.

### The `StringStore`
The `StringStore` acts as a lookup table that works in both directions:
- `String` -> `Hash`
- `Hash` -> `String`

The `StringStore` is shared across the entire `nlp` object, meaning all documents processed by that object share the same vocabulary.


````python
# Let's look at the hashing in action
coffee_hash = nlp.vocab.strings["coffee"]
coffee_string = nlp.vocab.strings[coffee_hash]

print(f"String: '{coffee_string}'")
print(f"Hash:   {coffee_hash}")

# It works for ANY string, even ones spaCy hasn't seen before
custom_hash = nlp.vocab.strings["Supercalifragilisticexpialidocious"]
print(f"\nCustom String Hash: {custom_hash}")
````

### The `Lexeme` Object

While a `Token` represents a word *in context* (e.g., the word "booking" could be a verb or a noun depending on the sentence), a **`Lexeme`** represents a word *out of context*.

A `Lexeme` contains information that is true for the word no matter where it appears (e.g., is it alphabetical? is it a number? what is its shape?). 

You get Lexemes directly from the `nlp.vocab`.


````python
# Let's get a Lexeme from the vocabulary
lexeme = nlp.vocab["coffee"]

print(f"Lexeme text: {lexeme.text}")
print(f"Lexeme hash: {lexeme.orth}")
print(f"Is alpha?  : {lexeme.is_alpha}")
print(f"Is title?  : {lexeme.is_title}")
print(f"Word Shape : {lexeme.shape_}")  # 'xxxx' means all lowercase letters
````

## 🔄 The Pipeline Data Flow

When you call `nlp(text)`, here is what happens sequentially:

1. **Tokenizer**: The raw text is split into a `Doc` of `Token` objects. (This is the only component that modifies the document structure).
2. **Tagger**: Assigns Part-of-Speech tags.
3. **Parser**: Assigns dependency labels and parses the syntactic structure. Also detects sentence boundaries.
4. **NER**: Detects and labels named entities.
5. **Lemmatizer**: Assigns base forms based on the POS tags.
6. **Attribute Ruler / Morphologizer**: Refines tags and morphological features.

*Note: The exact pipeline components depend on the model you load.*
